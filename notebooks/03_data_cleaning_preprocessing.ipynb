```python
from sklearn.exceptions import NotFittedError
from sklearn.utils.validation import check_is_fitted
from finaccess_eswatini.preprocessing import build_preprocessor

for model in ('model1', 'model2'):
    preprocessor = build_preprocessor(model)
    try:
        check_is_fitted(preprocessor)
    except NotFittedError:
        print(f'{model}: unfitted template confirmed')
    else:
        raise AssertionError(f'{model} preprocessor was unexpectedly fitted')
```
