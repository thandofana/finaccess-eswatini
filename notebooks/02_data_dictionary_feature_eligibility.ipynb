# Phase 2 — Data Dictionary & Feature Eligibility

This notebook reproduces the 199-variable dictionary and the independent, leakage-reviewed feature blueprints. It does not clean data or train models.

```python
from pathlib import Path
import csv
import json
import sys

project_root = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / 'pyproject.toml').is_file())
sys.path.insert(0, str(project_root / 'src'))

from finaccess_eswatini.phase2_dictionary import run

summary = run()
print(json.dumps({
    'variable_count': summary['variable_count'],
    'model1_candidate_count': summary['model1_candidate_count'],
    'model2_candidate_count': summary['model2_candidate_count'],
    'target_distributions': summary['target_distributions'],
}, indent=2))
```

```python
report_dir = project_root / 'reports' / 'phase_2'
def candidate_names(filename):
    with (report_dir / filename).open(encoding='utf-8', newline='') as source:
        return [row['variable'] for row in csv.DictReader(source)]

model1_features = candidate_names('candidate_features_model1.csv')
model2_features = candidate_names('candidate_features_model2.csv')
print('Model 1 features:', model1_features)
print('Model 2-only additions:', sorted(set(model2_features) - set(model1_features)))
```
