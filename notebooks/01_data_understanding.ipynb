# Phase 1 — Project Setup & Data Audit

This notebook is the portfolio entry point for the validated raw-data audit. It calls the reusable audit code and does not clean, transform, or model the microdata.

```python
from pathlib import Path
import csv
import json
import sys

project_root = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / 'pyproject.toml').is_file())
sys.path.insert(0, str(project_root / 'src'))

from finaccess_eswatini.data_audit import run

raw_path = project_root / 'data' / 'raw' / 'Findex_Microdata_2025_updateEswatini.csv'
report_dir = project_root / 'reports' / 'phase_1'
summary = run(raw_path, report_dir)
print(json.dumps({'shape': summary['shape'], 'targets': summary['targets'], 'duplicates': summary['duplicates']}, indent=2))
```

```python
with (report_dir / 'column_profile.csv').open(encoding='utf-8', newline='') as source:
    profiles = list(csv.DictReader(source))

highest_missingness = sorted(profiles, key=lambda row: float(row['missing_pct']), reverse=True)[:15]
for row in highest_missingness:
    print(f"{row['column']:<24} {float(row['missing_pct']):6.2f}% missing | {row['inferred_type']}")
```
